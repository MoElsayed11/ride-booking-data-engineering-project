```python
df = spark.read.table("ncr_ride_bookings")
```

```python
df = df.toDF(
    "date",
    "time",
    "booking_id",
    "booking_status",
    "customer_id",
    "vehicle_type",
    "pickup_location",
    "drop_location",
    "avg_vtat",
    "avg_ctat",
    "cancelled_by_customer",
    "customer_cancel_reason",
    "cancelled_by_driver",
    "driver_cancel_reason",
    "incomplete_rides",
    "incomplete_reason",
    "booking_value",
    "ride_distance",
    "driver_rating",
    "customer_rating",
    "payment_method"
)
```

```python
from pyspark.sql.functions import current_timestamp

df = df.withColumn("ingestion_time", current_timestamp())
```

```python
df.write.format("delta") \
    .mode("overwrite") \
    .saveAsTable("bronze_rides")
```
